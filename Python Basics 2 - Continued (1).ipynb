```python
# 5. Loops

# Loops are used to execute a block of code repeatedly. 
# The two main types of loops in Python are:

```

```python
#•	For Loop: Iterates over a sequence (like a list or string)
for i in range(5):
    print(i)
```

```python
#•	While Loop: Continues until a specified condition is false.
count = 0

while count < 5:
    print(count)
#    count += 1
    count = count + 1

```

```python
# Increment a number by 7 until it exceeds 30
x = 12
while x <= 30:
    x = x + 7
    print(x)
```

```python
# Increments a number by 7 and exit the loop before it exceeds 30

x = 12
while x <= 30:
    if (x+7) > 30:
        break
    x = x + 7
    print(x)
```

```python
#Or change the while condition so that x will not increment beyond 30

x = 12;
while (x+7) <= 30:
    x = x + 7
    print(x)
```

```python
# Python's enumerate() function

# Iterate over both the values and indices of all items in an iterable
# Use Python's enumerate() function
# Generates an (index, value) pair for each item in the iterable


# Key Features

# Index-Value Pairing: enumerate() pairs each item in the iterable with its corresponding index.
# Custom Starting Index: You can specify a starting index using the start parameter, which defaults to 0.
```

```python
seasons = ['Spring', 'Summer', 'Fall', 'Winter']

for index, season in enumerate(seasons, start = 0):
    print(f"Index: {index}, Season: {season}")
```

```python
fruits = ["apple", "banana", "cherry"]

for index, fruit in enumerate(fruits, start = 0):
 #   print(index, fruit)
    print(f"Index:{index}, Fruit:{fruit}")
```

```python
for number, fruit in enumerate(fruits, start=1):
    print(number, fruit)
```

```python
#It also works with strings and other iterables:

for index, letter in enumerate("lion", start = 0):
#    print(index, letter)
    print(f"Index: {index},  Letter:{letter} ")
```

```python
colors = ["red", "blue", "green"]

for number, color in enumerate(colors, start = 0):
    print(number, color)
```

```python
import time

t = int(input("Enter timer value in seconds: "))

while t > 0:
    print(t)
    time.sleep(1) # waiting tiime
    t -= 1

print("Time's up!")
```
